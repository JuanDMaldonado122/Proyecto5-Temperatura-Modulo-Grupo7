1. Situación de la energía solar en el SIN

La expansión fotovoltaica colombiana debe describirse distinguiendo capacidad efectiva neta en operación comercial, proyectos en pruebas y proyectos registrados o con posibilidad de conexión. Según XM, al 31 de diciembre de 2025 la capacidad efectiva neta total del Sistema Interconectado Nacional (SIN) fue de 21 028,56 MW; durante 2025 ingresaron 333,87 MW de nuevas plantas solares fotovoltaicas y 68 de los 75 proyectos de generación que entraron en operación fueron solares [1]. El mismo reporte de XM, difundido con el desglose tecnológico de la matriz, sitúa la capacidad solar fotovoltaica acumulada del SIN en 1 594,08 MW, aproximadamente 7,6 % del total instalado [1].

El crecimiento reciente ha sido muy rápido, aunque no debe interpretarse como una suma perfectamente lineal de todos los proyectos anunciados. En 2024 XM reportó la entrada de 1 379,9 MW solares dentro de los 1 446,6 MW de nueva generación de ese año [2]. En 2025 la adición fue menor, 333,87 MW, pero siguió siendo la principal fuente de nueva potencia. La diferencia entre adiciones anuales y el inventario acumulado puede obedecer a cambios de estado regulatorio, plantas en pruebas, ajustes de parámetros y fechas diferentes de corte; por eso, para un informe debe usarse como cifra principal el inventario de capacidad efectiva neta a una fecha determinada, y como indicador de expansión las adiciones anuales reportadas por XM [1], [2].

2. Irradiación y temperatura ambiente en Cali

El IDEAM reconoce el Atlas de radiación solar, ultravioleta y ozono de Colombia como uno de sus productos oficiales de meteorología [3], [4]. El Atlas representa un recurso espacial y climático: sus mapas no equivalen a una medición instantánea de la irradiancia sobre un módulo concreto. Para el diseño se debe especificar si se usa irradiación global horizontal (GHI), irradiación sobre el plano del generador (POA), promedio mensual, promedio anual o un valor horario.

La consulta pública recuperada para el Atlas no expuso una tabla descargable con el valor puntual del píxel de Cali. Por tanto, el dato defendible para una introducción es un rango preliminar, no una falsa precisión: para la región Andina se reportan aproximadamente 3,5–5,0 kWh/m²·día, y para Cali se suele utilizar como orden de magnitud cerca de 4,5 kWh/m²·día en estudios de prefactibilidad [5]. Este valor debe verificarse para las coordenadas exactas del proyecto con el geovisor o la base de datos del IDEAM, porque puede cambiar con la ubicación, orientación, nubosidad, periodo climático y variable empleada.

La temperatura ambiente también debe identificarse por estación y periodo. El portal del IDEAM proporciona el acceso a datos meteorológicos y normales climatológicas [3]. Para un cálculo preliminar en Cali puede adoptarse una temperatura ambiente media cercana a 24 °C, pero debe sustituirse en el diseño definitivo por la temperatura de la estación IDEAM más representativa y, para análisis térmico, por la serie horaria o por el percentil de temperatura de diseño. No debe confundirse la temperatura ambiente con la temperatura de la celda: bajo irradiancia elevada, la celda puede estar decenas de grados por encima del aire.

Como control independiente de conceptos, el Global Solar Atlas define GHI en kWh/m² como promedio anual y temperatura del aire a 2 m en °C, mostrando que irradiación y temperatura son variables distintas y deben descargarse con sus respectivas unidades [6].

3. Física térmica del módulo, NOCT y modelo de Ross

Un módulo se calienta más que el aire porque la irradiación absorbida por el vidrio, las celdas y las capas posteriores se reparte entre tres destinos: electricidad útil, radiación térmica y calor transferido por conducción y convección. Una celda de silicio convierte solo una fracción de la irradiancia incidente; el resto termina principalmente como calor. El equilibrio estacionario se alcanza cuando el calor absorbido iguala las pérdidas por convección hacia el aire, radiación hacia el cielo y el entorno, y conducción hacia la estructura. Si la convección es débil —poco viento o poco espacio posterior— la temperatura de la celda aumenta.

La NOCT (Nominal Operating Cell Temperature) es un parámetro de comparación térmica del módulo, no una temperatura máxima ni una predicción universal para cualquier instalación. La definición convencional se basa aproximadamente en irradiancia de 800 W/m², temperatura ambiente de 20 °C, viento de 1 m/s y montaje abierto que permite la ventilación posterior; la condición de ensayo incluye el estado eléctrico especificado por el procedimiento aplicable. En módulos cristalinos modernos son frecuentes valores de alrededor de 43–47 °C, pero se debe usar el valor de la ficha técnica del modelo instalado y comprobar si el fabricante reporta NOCT o NMOT, ya que no son siempre idénticos [7].

El modelo lineal de Ross es una aproximación útil:

$$[ T_c = T_a + kG, ]$$

donde (T_c) es la temperatura de celda, (T_a) la temperatura ambiente, (G) la irradiancia en el plano del módulo y (k) un coeficiente térmico en °C/(W/m²). Si se obtiene (k) a partir de una NOCT convencional, una forma habitual es:

$$[ k \approx \frac{NOCT-20}{800}. ]$$

Por ejemplo, con NOCT = 45 °C, (k\approx0,0313) °C/(W/m²). Con (T_a=24) °C y (G=800) W/m², el modelo estima ($T_c\approx49$) °C. Es una estimación de primer orden: Ross no representa explícitamente la velocidad del viento, la geometría del bastidor, el intercambio radiativo con el cielo ni una cámara posterior confinada. Por eso, su coeficiente debe calibrarse cuando el módulo está integrado en una fachada, pegado a una cubierta o instalado en condiciones de ventilación distintas. Un estudio experimental de sistemas BIPV que comparó Ross y Faiman encontró que Ross podía ser adecuado para predecir energía anual, pero subrayó la importancia de las condiciones del dorso y del ajuste con datos medidos en el sitio [7].

4. Efecto de la temperatura sobre la potencia

El coeficiente de temperatura de la potencia máxima, ($\gamma_{Pmax}$), expresa el cambio relativo de la potencia máxima por cada grado Celsius de variación de la temperatura de la celda, normalmente respecto de las condiciones estándar de ensayo (STC: 1000 W/m², espectro AM1.5 y temperatura de celda de 25 °C). Para silicio cristalino el coeficiente es negativo: al aumentar la temperatura disminuye principalmente el voltaje de operación. Una ficha técnica oficial de LONGi reporta ($\gamma_{Pmax}=-0,280\ %/°C$) para el modelo consultado [8]. En ingeniería preliminar, valores del orden de −0,25 a −0,35 %/°C son un rango razonable para muchos módulos de silicio cristalino, pero la ficha del modelo debe prevalecer.

La aproximación lineal es:

$$ P(T_c)\approx P_{STC}\left[1+\frac{\gamma_{Pmax}}{100}(T_c-25)\right]$$

Así, con −0,280 %/°C, cada aumento de 1 °C sobre 25 °C reduce aproximadamente 0,28 % de la potencia respecto de la potencia a 25 °C, manteniendo constantes irradiancia y demás condiciones. Si la celda opera a 65 °C, el incremento es de 40 °C y la pérdida térmica aproximada es 11,2 %. Esta cifra no es una pérdida energética anual automática: la producción anual también depende de irradiancia, espectro, suciedad, sombras, inversor, mismatch, disponibilidad y temperatura variable durante el día.

5. Hot spots, diagnóstico, viento y montaje

Un hot spot es una región localizada del módulo con temperatura anormalmente alta. Puede aparecer cuando una celda sombreada, agrietada, sucia o eléctricamente desajustada queda forzada a conducir en polarización inversa dentro de una cadena serie; la potencia de otras celdas puede disiparse localmente en esa zona. La consecuencia puede ser pérdida de potencia, degradación acelerada del encapsulante, daño de soldaduras o, en casos severos, riesgo de daño térmico. Los diodos bypass limitan algunos efectos del sombreado, pero no eliminan todas las causas de puntos calientes.

La detección debe combinar métodos:

Termografía infrarroja (IR/IRT): identifica diferencias de temperatura entre celdas, diodos, conexiones y módulos durante operación. Es eficaz como cribado de grandes plantas, incluso con drones, pero depende de irradiancia suficiente, viento, ángulo de observación, emisividad, calibración y temperatura ambiente. Una revisión reciente concluye que la termografía es principalmente una herramienta cualitativa o semicuantitativa y no debe convertirse por sí sola en un criterio universal de aceptación o rechazo [9].
Inspección visual: permite localizar suciedad, delaminación, decoloración, roturas, vidrio dañado y sombras persistentes.
Electroluminiscencia (EL): útil para microfisuras y defectos de células, normalmente en condiciones controladas y con el módulo polarizado eléctricamente.
Curvas I–V, medición de aislamiento y datos del inversor/SCADA: ayudan a confirmar si una anomalía térmica produce una pérdida eléctrica y a separar una falla de módulo de un problema de cadena.
El viento enfría el módulo al aumentar la convección; en consecuencia, a igual irradiancia y temperatura ambiente, una velocidad de viento mayor suele reducir (T_c). El modelo de Ross incorpora este efecto solo de forma implícita mediante su coeficiente (k), mientras que modelos como Faiman lo representan mediante términos dependientes de la velocidad del viento. El tipo de montaje modifica la superficie de intercambio y la renovación de aire: un bastidor abierto con separación posterior suele disipar más calor que un módulo adosado directamente a una cubierta o integrado en una fachada. En BIPV, la cámara posterior, sus aberturas y la dirección del flujo pueden cambiar sustancialmente la temperatura; no es correcto imponer universalmente una penalización fija de 10–15 °C sin medir o modelar la configuración concreta [7].

Conclusión para la introducción

En Colombia, la fotovoltaica ya representa una fracción relevante del SIN y fue la principal fuente de nueva capacidad durante 2024–2025 [1], [2]. En Cali, los valores de recurso deben presentarse como climatología y rango espacial —no como una medición instantánea— y verificarse para las coordenadas del proyecto en el Atlas y las estaciones del IDEAM [3]–[6]. La irradiancia eleva la temperatura de la celda por encima del aire; Ross y NOCT permiten una primera estimación, mientras que el viento y el montaje determinan el error de esa aproximación. Finalmente, el coeficiente de (P_{max}) traduce el calentamiento en una pérdida típica cercana a 0,28 % por °C para el módulo LONGi consultado [8], y la termografía combinada con inspección visual, EL y mediciones eléctricas constituye una estrategia más confiable para diagnosticar hot spots [9].
